```python
import pandas as pd
```

```python
# T1

# Load Data
data = pd.read_csv("titanic-data.csv")
```

```python
# Basic Info
print(data.shape, data.size, data.columns, data.info())
```

```python
# Descriptives
print(data.describe())
print(data.describe(include=['object']))
```

```python
# Delete PassengerId
data = data.drop(columns=["PassengerId"])
print(data.describe())
```

```python
# Count Missing Values And Fill In New Ones
print(data.isna().sum())
data["Age"] = data["Age"].fillna(data["Age"].mean())
data["Embarked"] = data["Embarked"].fillna(data["Embarked"].mode()[0])
```

```python
# Encode Sex to 0/1
data["Sex01"] = data["Sex"].map({"male":1, "female":0})
print(data["Sex01"].unique())
```

```python
# Cor of Survived
print(data.corr(numeric_only=True)["Survived"])
```

```python
# Sort By Fare Price
data = data.sort_values(by="Fare")
print(data[["Fare"]].head(10))
print(data[["Fare"]].tail(10))
```

```python
# Split By "."
data["Name_before_dot"] = data["Name"].apply(lambda x: x.split(".")[0])
data["Name_after_dot"] = data["Name"].apply(lambda x: x.split(".")[1].strip() if "." in x else None)
print(data[["Name", "Name_before_dot", "Name_after_dot"]].head(10))
```

```python
# Extract Numeric Part of Ticket By Blank
data["Ticket_num"] = data["Ticket"].apply(lambda x: [part for part in x.split() if part.isdigit()][-1] 
                                          if any(part.isdigit() for part in x.split()) else "")
print(data[["Ticket_num"]].head(10))
```

```python
# Embark: S to 1
data["Embarked"] = data["Embarked"].str.replace("S", "1")
print(data[["Embarked"]].head(10))
print(data["Embarked"].unique())
```

```python
# T2

# Load Data
trade = pd.read_csv("trade.csv", encoding="gbk")
print(trade.info())
```

```python
# Generate Year & Month
trade["year"] = trade["数据年月"].astype(str).str[:4].astype(int)
trade["month"] = trade["数据年月"].astype(str).str[4:6].astype(int)
print(trade[["数据年月", "year", "month"]].head(10))
print(trade[["数据年月", "year", "month"]].tail(10))
```

```python
# Rid of ","
trade["人民币"] = trade["人民币"].str.replace(",", "")
print(trade[["人民币"]].head(10))
```

```python
# To Numeric
print(trade["人民币"].dtype)
trade["人民币"] = pd.to_numeric(trade["人民币"])
print(trade["人民币"].dtype)
```

```python
# Sum Up RMB By Trade Partners
print(trade.groupby("贸易伙伴名称")["人民币"].sum())
```

```python
# Pivot Table
pivot = pd.pivot_table(
    trade,
    index=["贸易伙伴名称", "贸易方式名称"],
    values=["人民币", "第一数量", "第二数量"],
    aggfunc="sum"
)
print(pivot)
```

```python
# T3

# Load Data
m = pd.read_csv("maotai.csv", encoding="utf-8")

# Date to Datetime
m["日期"] = pd.to_datetime(m["日期"])

# Generate Year & Month
m["year"] = m["日期"].dt.year
m["month"] = m["日期"].dt.month

# Monthly Mean
monthly_mean = m.groupby(["year", "month"])["收盘价"].mean()
print(monthly_mean)

# Pivot Table
pivot_m = m.pivot_table(
    index="year",
    columns="month",
    values="收盘价",
    aggfunc="mean"
)
print(pivot_m)
```
